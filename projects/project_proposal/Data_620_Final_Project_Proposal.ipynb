### **Project Purpose**

Our project aims to understand the ecosystem of professional sports during the trade season, focusing specifically on the NBA. By mapping trades as a network, we want to discover which teams act as central hubs for talent exchange and analyze how media narrative and fan sentiment correlate with actual decisions.

<br>

---

<br>

### **Plan**

The first step is to create a comprehensive transaction log of trades. This will include the player's name, origin team, destination team, trade year, and important performance and financial metrics like salary, contract length, and average points per game (PPG). We will compile this information using CSV files and utilize pandas to convert the data into structured dataframes for analysis.

Next, we’ll construct the network graph using networkx and matplotlib. In this model, teams will represent the nodes, and the trades will form the directed edges between them. We plan to measure:
- Degree Centrality: To determine which teams handle the highest volume of trades.
- Betweenness Centrality: To identify "bridge" teams that facilitate complex multi-team trades.
- Edge Weights: We will use thicker lines to represent higher player salaries, helping us visualize how financial constraints drive trade behaviors or free-agency decisions.

Finally, we will gather public commentary and sports articles surrounding the trade deadline to analyze public sentiment regarding specific players and teams. Using Natural Language Processing, we will extract overarching themes (such as "rebuilding phases" or "win-now pressure") to see how public discourse aligns with the actual trades that take place.

### **Concerns**

API and Scraping Restrictions: Gathering text data from platforms like X or Reddit can run into strict rate limits, so we'd supplement our data with public sports feeds.

Complex Multi-Team Trades: The NBA frequently features 3 or 4 team trades, which do not inherently fit a simple A-to-B network path. Therefore, we'd have to break up years into pairs.

Multi-Year Contracts: Long-term contracts can blur the snapshot of a single trade deadline. Similarly, we'd have to break up the years into seasonal pairs.
