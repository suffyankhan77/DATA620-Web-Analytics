## Project Purpose

Our project aims to examine the NBA trade ecosystem during the 2025–2026 season
and the 2026 offseason through a clearly documented data-collection cutoff date. By representing completed player movements as a directed network, we plan to identify teams that serve as central hubs for talent and
financial exchange. We will also analyze sports-media narratives and public
sentiment surrounding selected major trades and compare those narratives with
the teams and transactions identified as structurally important in the network.

<br>

## Motivation and Guiding Research Questions

NBA trades are often evaluated as isolated transactions involving individual
players and teams. However, these transactions collectively form a league-wide
network of repeated talent and financial exchanges. Network analysis can reveal
which teams occupy influential positions within this system, while text analysis
can show how fans and sports media interpret the same transactions. Combining
these perspectives may provide a more complete understanding of NBA trade
activity than examining trade records or public commentary independently.

The project will be guided by the following research questions:

1. Which NBA teams are the most central in the 2025–2026 season and 2026
   offseason trade network, through the project’s documented cutoff date,
   based on in-degree, out-degree, weighted in-degree, weighted out-degree,
   and betweenness centrality?

2. How do transaction volume and player salary movement relate to a team’s
   structural position within the trade network?

3. What sentiments and recurring themes appear in public commentary and sports
   coverage surrounding selected major NBA trades?

4. How do these public narratives compare with the teams and transactions
   identified as important through the network analysis?

<br>


## Plan

The first step will be to create a comprehensive transaction log of completed NBA trades. The dataset will include the player’s name, origin team, destination team, trade date, season, salary, remaining contract length, and relevant performance measures such as points per game. We will compile the information into CSV files and use `pandas` to clean, combine, and structure the data for analysis.

Next, we will construct a directed network using `NetworkX` and visualize it with `Matplotlib`. NBA teams will represent the nodes, and each player movement will create a directed edge from the team sending the player to the team receiving the player. Each player movement will first be stored as a separate record in the network edge list. For team-level analysis, movements between the same origin and destination teams may be aggregated, while a shared trade identifier will
preserve the connection between movements belonging to the same transaction.

We plan to measure:

- **In-Degree and Out-Degree Centrality:** To distinguish teams receiving the
  most player movements from teams sending the most player movements.

- **Weighted In-Degree and Out-Degree:** To measure incoming and outgoing trade
  activity using two alternative edge weights: the number of player movements
  and the total salary value associated with those movements.

- **Betweenness Centrality:** To identify teams that occupy important structural
  positions within the trade network.

- **Edge Visualization:** To display transaction volume or salary value through
  edge thickness, with thicker edges representing greater trade activity or
  financial value.

Finally, we will gather publicly available sports articles and public commentary surrounding selected major trades. Using natural language processing, we will clean and tokenize the text, analyze sentiment toward specific teams and players,
and identify recurring themes such as rebuilding, championship contention, salary-cap pressure, and “win-now” strategies. We will then compare these narratives with the teams and transactions identified as important through the network analysis. Depending on the size and accessibility of the final corpus, sentiment and thematic results may be evaluated through manual review of a sample of documents and inspection of representative keywords, comments, and articles.

## Concerns

### API and Scraping Restrictions

Accessing text from platforms such as X and Reddit may involve API limitations, rate limits, or reproducibility concerns. Therefore, publicly accessible sports-news articles and available Reddit content will serve as the primary text sources, while X will be treated as an optional source.

### Complex Multi-Team Trades

Three-team and four-team trades cannot always be represented as a single team-to-team edge. We will decompose each transaction into directed player-movement edges while preserving a shared trade identifier so that all movements remain connected to the original transaction.

### Multi-Year Contracts

Player contracts may extend beyond the season in which a trade occurs. We will record salary and remaining contract length as attributes at the time of the transaction rather than treating each contract year as a separate trade.

### Data Consistency

Trade dates, team names, salaries, and player statistics may differ across sources. We will standardize team names, verify major transactions using multiple sources, and document any missing or estimated values.

### Text-Corpus Scope

Collecting commentary for every trade may be too broad. The text analysis may therefore focus on a selected group of major trades based on salary value, media attention, or network importance.

Because this proposal is a planning document, the exact number of trades,
text sources, collection period, and natural language processing methods may
be refined based on data accessibility, quality, and the time available for
the final analysis.
